# Bài 5. Reshape và Flatten

```python
import torch
a = torch.randn(3, 4)
print("Original:")
print(a)
print("\nFlatten:")
print(a.ravel())
print("\nReshape to 3 x 2 x 2:")
print(a.reshape(3, 2, 2))
```

### yêu cầu

#### Kiểm tra số phần tử trước và sau khi reshape.
#### Thử reshape Tensor thành kích thước 2 × 6.

```python
so_phan_tu_truoc = a.numel()
a_2x6 = a.reshape(2, 6)
so_phan_tu_sau = a_2x6.numel()

print("Số phần tử trước khi reshape:", so_phan_tu_truoc)
print("Tensor sau khi reshape thành 2 x 6:")
print(a_2x6)
print("Shape sau khi reshape:", a_2x6.shape)
print("Số phần tử sau khi reshape:", so_phan_tu_sau)
print("Số phần tử không thay đổi:", so_phan_tu_truoc == so_phan_tu_sau)
```

# Bài 6. Phép toán và nhân ma trận

```python
import torch
a = torch.randn(2, 3)
b = torch.randn(2, 3)
print("a:")
print(a)
print("b:")
print(b)
print("\na + b:")
print(a + b)
print("\na / b:")
print(a / b)
print("\na^2:")
print(a ** 2)
print("\nMatrix multiplication:")
print(a @ b.T)
```

# Bài 7. Các hàm thống kê trên Tensor

```python
import torch
a = torch.randn(3, 4)
print(a)
print("\nMean:")
print(torch.mean(a, dim=0))
print("\nStandard deviation:")
print(torch.std(a, dim=0))
print("\nCumulative sum:")
print(torch.cumsum(a, dim=0))
```

### yêu cầu

#### Thay dim=0 bằng dim=1 và nhận xét sự khác nhau.

```python
a = torch.randn(3, 4)

print("Tensor a:")
print(a)

mean_dim1 = torch.mean(a, dim=1)
std_dim1 = torch.std(a, dim=1)
cumsum_dim1 = torch.cumsum(a, dim=1)

print("\nMean theo dim=1:")
print(mean_dim1)
print("Shape:", mean_dim1.shape)

print("\nStandard deviation theo dim=1:")
print(std_dim1)
print("Shape:", std_dim1.shape)

print("\nCumulative sum theo dim=1:")
print(cumsum_dim1)
print("Shape:", cumsum_dim1.shape)
```

**Nhận xét:** Khi dùng `dim=0`, các phép tính được thực hiện theo từng cột nên `mean` và `std` trả về 4 giá trị. Khi dùng `dim=1`, các phép tính được thực hiện theo từng hàng nên `mean` và `std` trả về 3 giá trị. `cumsum(dim=1)` tính tổng tích lũy từ trái sang phải trong từng hàng và vẫn giữ nguyên shape `(3, 4)`.

# Bài 8. Autograd cơ bản

```python
import torch
x = torch.tensor(
3.6,
requires_grad=True
)
y = x * x
y.backward()
print("x =", x)
print("y =", y)
print("x.grad =", x.grad)
```

### yêu cầu

#### Tính thủ công đạo hàm của y = x2 tại x = 3.6 và so sánh với x.grad.
#### Đổi hàm thành y = 3*x**2 + 2*x + 1 và kiểm tra gradient.

```python
x = torch.tensor(3.6, requires_grad=True)
y = x ** 2
y.backward()

gradient_thu_cong_1 = 2 * 3.6
print("Hàm y = x^2")
print("Gradient bằng Autograd:", x.grad.item())
print("Gradient tính thủ công:", gradient_thu_cong_1)
print("Kết quả khớp:", torch.isclose(x.grad, torch.tensor(gradient_thu_cong_1)).item())

x_moi = torch.tensor(3.6, requires_grad=True)
y_moi = 3 * x_moi ** 2 + 2 * x_moi + 1
y_moi.backward()

gradient_thu_cong_2 = 6 * 3.6 + 2
print("\nHàm y = 3*x^2 + 2*x + 1")
print("Gradient bằng Autograd:", x_moi.grad.item())
print("Gradient tính thủ công:", gradient_thu_cong_2)
print("Kết quả khớp:", torch.isclose(x_moi.grad, torch.tensor(gradient_thu_cong_2)).item())
```

# Bài 9. Gradient Descent tìm hệ số đa thức

```python
import numpy as np
import torch
polynomial = np.poly1d([1, 2, 3])
N = 20
X = np.random.randn(N, 1) * 5
Y = polynomial(X)
XX = np.hstack([
X * X,
X,
np.ones_like(X)
])
w = torch.randn(3, 1, requires_grad=True)
x = torch.tensor(XX, dtype=torch.float32)
y = torch.tensor(Y, dtype=torch.float32)
optimizer = torch.optim.NAdam([w], lr=0.01)
print("Initial coefficients:")
print(w)
for epoch in range(1000):
    optimizer.zero_grad()
    y_pred = x @ w
    mse = torch.mean(torch.square(y - y_pred))
    mse.backward()
    optimizer.step()
print("Learned coefficients:")
print(w)
```

### yêu cầu

#### So sánh ba hệ số học được với [1, 2, 3].
#### Giải thích vai trò của optimizer.zero_grad(), mse.backward() và optimizer.step().

```python
he_so_hoc_duoc = w.detach().flatten()
he_so_mong_doi = torch.tensor([1.0, 2.0, 3.0])
sai_so = he_so_hoc_duoc - he_so_mong_doi

print("Ba hệ số học được [x^2, x, hằng số] =", he_so_hoc_duoc)
print("Ba hệ số mong đợi =", he_so_mong_doi)
print("Sai số từng hệ số =", sai_so)
print("Sai số tuyệt đối trung bình =", torch.mean(torch.abs(sai_so)).item())
```

# Bài 10. Giải hệ phương trình bằng Autograd

```python
import random
import torch
A_bai10 = torch.tensor(random.random(), requires_grad=True)
B_bai10 = torch.tensor(random.random(), requires_grad=True)
C_bai10 = torch.tensor(random.random(), requires_grad=True)
D_bai10 = torch.tensor(random.random(), requires_grad=True)
EPOCHS = 2000
optimizer = torch.optim.NAdam([A_bai10, B_bai10, C_bai10, D_bai10], lr=0.01)
for _ in range(EPOCHS):
    y1 = A_bai10 + B_bai10 - 9
    y2 = C_bai10 - D_bai10 - 1
    y3 = A_bai10 + C_bai10 - 8
    y4 = B_bai10 - D_bai10 - 2
    sqerr = y1*y1 + y2*y2 + y3*y3 + y4*y4
    optimizer.zero_grad()
    sqerr.backward()
    optimizer.step()
print("A =", A_bai10)
print("B =", B_bai10)
print("C =", C_bai10)
print("D =", D_bai10)
```

### yêu cầu

#### Kiểm tra nghiệm thu được với bốn phương trình: A+B=9, C-D=1, A+C=8, B-D=2.
#### Nhận xét vì sao bài toán có thể có nhiều nghiệm.

```python
A_value = A_bai10.detach().item()
B_value = B_bai10.detach().item()
C_value = C_bai10.detach().item()
D_value = D_bai10.detach().item()

residuals = {
    "A + B - 9": A_value + B_value - 9,
    "C - D - 1": C_value - D_value - 1,
    "A + C - 8": A_value + C_value - 8,
    "B - D - 2": B_value - D_value - 2,
}

print("Sai số của bốn phương trình:")
for equation, residual in residuals.items():
    print(f"{equation} = {residual:.6f}")

print("\nNghiệm tổng quát với t = D:")
print("A = 7 - t")
print("B = t + 2")
print("C = t + 1")
print("D = t")

for t in [0.0, 1.0, 2.0]:
    solution = (7 - t, t + 2, t + 1, t)
    print(f"t = {t}: (A, B, C, D) = {solution}")
```

```python
import numpy as np

polynomial = np.poly1d([2, -3, 5])
N = 100
X = np.random.randn(N, 1) * 5
Y = polynomial(X)
XX = np.hstack([X * X, X, np.ones_like(X)])

w = torch.randn(3, 1, requires_grad=True)
features = torch.tensor(XX, dtype=torch.float32)
targets = torch.tensor(Y, dtype=torch.float32)
optimizer = torch.optim.NAdam([w], lr=0.01)

for _ in range(10000):
    optimizer.zero_grad()
    prediction = features @ w
    mse = torch.mean((targets - prediction) ** 2)
    mse.backward()
    optimizer.step()

print("Hệ số tìm được [x^2, x, hằng số] =", w.detach().flatten())
print("Hệ số mong đợi [2, -3, 5] =", torch.tensor([2.0, -3.0, 5.0]))
print("MSE cuối =", mse.item())
```
