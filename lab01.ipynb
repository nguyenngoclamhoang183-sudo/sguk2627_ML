# Bài 9. Gradient Descent tìm hệ số đa thức

```python
import numpy as np
import torch
polynomial = np.poly1d([1, 2, 3])
N = 20
X = np.random.randn(N, 1) * 5
Y = polynomial(X)
XX = np.hstack([
X * X,
X,
np.ones_like(X)
])
w = torch.randn(3, 1, requires_grad=True)
x = torch.tensor(XX, dtype=torch.float32)
y = torch.tensor(Y, dtype=torch.float32)
optimizer = torch.optim.NAdam([w], lr=0.01)
print("Initial coefficients:")
print(w)
for epoch in range(1000):
    optimizer.zero_grad()
    y_pred = x @ w
    mse = torch.mean(torch.square(y - y_pred))
    mse.backward()
    optimizer.step()
print("Learned coefficients:")
print(w)
```

# Bài 10. Giải hệ phương trình bằng Autograd

```python
import random
import torch
A = torch.tensor(random.random(), requires_grad=True)
B = torch.tensor(random.random(), requires_grad=True)
C = torch.tensor(random.random(), requires_grad=True)
D = torch.tensor(random.random(), requires_grad=True)
EPOCHS = 2000
optimizer = torch.optim.NAdam([A, B, C, D], lr=0.01)
for _ in range(EPOCHS):
    y1 = A + B - 9
    y2 = C - D - 1
    y3 = A + C - 8
    y4 = B - D - 2
    sqerr = y1*y1 + y2*y2 + y3*y3 + y4*y4
    optimizer.zero_grad()
    sqerr.backward()
    optimizer.step()
print("A =", A)
print("B =", B)
print("C =", C)
print("D =", D)
```

```python
x = torch.tensor(2.0, requires_grad=True)
y = 2 * x**2 + 5 * x + 3
y.backward()

dy_dx_thu_cong = 4 * 2 + 5

print("y =", y.item())
print("dy/dx bằng autograd =", x.grad.item())
print("dy/dx thủ công =", dy_dx_thu_cong)
print("Kết quả khớp:", torch.isclose(x.grad, torch.tensor(float(dy_dx_thu_cong))).item())
```

```python
import numpy as np

polynomial = np.poly1d([2, -3, 5])
N = 100
X = np.random.randn(N, 1) * 5
Y = polynomial(X)
XX = np.hstack([X * X, X, np.ones_like(X)])

w = torch.randn(3, 1, requires_grad=True)
features = torch.tensor(XX, dtype=torch.float32)
targets = torch.tensor(Y, dtype=torch.float32)
optimizer = torch.optim.NAdam([w], lr=0.01)

for _ in range(10000):
    optimizer.zero_grad()
    prediction = features @ w
    mse = torch.mean((targets - prediction) ** 2)
    mse.backward()
    optimizer.step()

print("Hệ số tìm được [x^2, x, hằng số] =", w.detach().flatten())
print("Hệ số mong đợi [2, -3, 5] =", torch.tensor([2.0, -3.0, 5.0]))
print("MSE cuối =", mse.item())
```
